# Automatic Speech Recognition Experiment

## Whisper Baseline

### Objective

The objective of this experiment is to establish an initial Automatic Speech Recognition (ASR) baseline using OpenAI's Whisper model.

The pipeline is:

Audio Input → Whisper Model → Text Transcription

```python
import torch
import librosa
from transformers import pipeline

print("Libraries imported successfully!")
```

```python
model_name = "openai/whisper-base"

asr_pipeline = pipeline(
    "automatic-speech-recognition",
    model=model_name
)

print("Whisper model loaded successfully!")
```

```python
audio_file = "/content/voice.wav"

result = asr_pipeline(
    audio_file,
    return_timestamps=True
)

print("Transcription:")
print(result["text"])
```

## Observation

The Whisper Base model successfully generated a transcription from the input audio.

However, repetitive tokens were observed during the initial portion of the transcription. This may be caused by the presence of non-speech audio, background music, or long-form generation behaviour.

This experiment demonstrates the importance of evaluating ASR models on clean speech samples representative of the Aksharvel application.
